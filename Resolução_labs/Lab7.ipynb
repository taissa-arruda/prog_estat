**Exercício 1.1**

```python
import numpy as np
import matplotlib.pyplot as plt

n = 1000

def f(x):
    return 20 * x * (1 - x)**3

malha = np.linspace(0, 1, 10000)
M = np.max(f(malha))
print("M aproximado:", M)

amostra = []
tentativas = []

for i in range(n):
    contador = 0
    aceitou = False
    while aceitou == False:
        y = np.random.uniform(0, 1)
        u = np.random.uniform(0, 1)
        contador = contador + 1
        if u <= f(y) / M:
            amostra.append(y)
            tentativas.append(contador)
            aceitou = True
```

**Exercício 1.2**

```python
plt.hist(amostra, bins=50, density=True, label="Histograma da amostra")
plt.plot(malha, f(malha), color="red", label="Densidade teórica f(x)")
plt.xlabel("x")
plt.ylabel("Densidade")
plt.legend()
plt.show()
```

**Exercício 1.3**

```python
p = 1 / M

print("Probabilidade teórica de aceitação:", p)
print("Proporção empírica de aceitação:", n / np.sum(tentativas))
print("Número esperado de tentativas (teórico):", M)
print("Média empírica do número de tentativas:", np.mean(tentativas))
```

```python
valores_k = []
prob_empirica = []
prob_teorica = []

for k in range(1, 11):
    valores_k.append(k)
    prob_empirica.append(tentativas.count(k) / n)
    prob_teorica.append((1 - p)**(k - 1) * p)

for i in range(10):
    print("k =", valores_k[i], "| empírica:", prob_empirica[i], "| teórica:", prob_teorica[i])

plt.bar(valores_k, prob_empirica, label="Empírica")
plt.plot(valores_k, prob_teorica, "o-", color="red", label="Teórica (Geométrica)")
plt.xlabel("Número de tentativas")
plt.ylabel("Probabilidade")
plt.legend()
plt.show()
```
