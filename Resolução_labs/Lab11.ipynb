```python
import numpy as np

np.random.seed(0)
n = 10**4
```

**Exercício 1**

```python
# Gamma(3, 2) = soma de 3 exponenciais Exp(2)
u = np.random.uniform(0, 1, size=(n, 3))   # n linhas, 3 colunas
e = -np.log(u) / 2                         # Exp(2) por inversão
x = np.sum(e, axis=1)                      # soma de cada linha
h = np.sqrt(x)

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 1.174982")
```

**Exercício 2**

```python
# Beta(2, 5) = G1/(G1+G2), com G1 ~ Gamma(2, 1) e G2 ~ Gamma(5, 1)
g1 = np.sum(-np.log(np.random.uniform(0, 1, size=(n, 2))), axis=1)
g2 = np.sum(-np.log(np.random.uniform(0, 1, size=(n, 5))), axis=1)
x = g1 / (g1 + g2)
h = (x > 0.4)

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 0.23328")
```

**Exercício 3**

```python
# Normal padrão por Box-Muller
u = np.random.uniform(0, 2*np.pi, n)
t = -np.log(np.random.uniform(0, 1, n))    # T ~ Exp(1)
z = np.sqrt(2*t) * np.cos(u)               # Z ~ N(0, 1)
x = 1 + 2*z                                # X ~ N(1, 4): média 1, desvio padrão 2
h = x**2 + 2*x + 1

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 8")
```

**Exercício 4**

```python
x = np.sum(-np.log(np.random.uniform(0, 1, size=(n, 2))), axis=1)   # Gamma(2, 1)
y = -np.log(np.random.uniform(0, 1, n))                             # Exp(1)
h = (x > y)

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 0.75")
```

**Exercício 5**

```python
# Beta(3, 4) = G1/(G1+G2), com G1 ~ Gamma(3, 1) e G2 ~ Gamma(4, 1)
g1 = np.sum(-np.log(np.random.uniform(0, 1, size=(n, 3))), axis=1)
g2 = np.sum(-np.log(np.random.uniform(0, 1, size=(n, 4))), axis=1)
x = g1 / (g1 + g2)
h = np.exp(x) / 60        # B(3,4) = 1/60

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 0.02598262")
```

**Exercício 6**

```python
# Box-Muller gera duas normais independentes de uma vez (cos e sen)
u = np.random.uniform(0, 2*np.pi, n)
t = -np.log(np.random.uniform(0, 1, n))
x = np.sqrt(2*t) * np.cos(u)
y = np.sqrt(2*t) * np.sin(u)
h = (x**2 + y**2 > 5)

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico (qui-quadrado com 2 g.l.):", np.exp(-5/2))
```

**Exercício 7**

```python
u = np.random.uniform(0, 1, size=(n, 5))
x = -np.log(u) / 2            # Exp(2) por inversão
h = np.max(x, axis=1)         # máximo de cada linha

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 1.141667")
```

**Exercício 8**

```python
u = np.random.uniform(0, 2*np.pi, n)
t = -np.log(np.random.uniform(0, 1, n))
z = np.sqrt(2*t) * np.cos(u)     # Z ~ N(0, 1) por Box-Muller
x = np.exp(z)

# Média
estimativa = np.mean(x)
s = np.std(x, ddof=1)
ep = s / np.sqrt(n)
print("mu estimado:", estimativa)
print("Variância do estimador:", s**2 / n)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico:", np.exp(1/2))

# Probabilidade
h = (x > 10)
estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
print("p estimado:", estimativa)
print("Variância do estimador:", s**2 / n)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
print("Valor teórico: 0.0106511")
```
