**Exercício 1**

```python
import numpy as np

np.random.seed(0)   # para os resultados se repetirem
n = 10**4           # número de simulações
```

```python
x = np.random.uniform(0, 1, n)   # 1. gerar
h = np.exp(-x**2)                # 2. calcular h(X)

estimativa = np.mean(h)          # 3. estimativa
s = np.std(h, ddof=1)            # 4. desvio padrão amostral (n-1)
ep = s / np.sqrt(n)              #    erro padrão
variancia = s**2 / n             #    variância do estimador

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)   # 5. IC
```

**Exercício 2**

```python
x = np.random.normal(0, 1, n)
h = (x > 1)          # vale 1 (True) se X > 1 e 0 (False) se não

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
```

**Exercício 3**

```python
x = np.random.uniform(0, 1, n)
y = np.random.uniform(0, 1, n)
h = 4 * (x**2 + y**2 <= 1)   # 4 se o ponto cai no círculo, 0 se não

estimativa = np.mean(h)
s = np.std(h, ddof=1)
ep = s / np.sqrt(n)
variancia = s**2 / n

print("Estimativa:", estimativa)
print("Variância do estimador:", variancia)
print("IC 95%:", estimativa - 1.96*ep, estimativa + 1.96*ep)
```

**Exercício 4**

```python
x = np.random.uniform(0, 1, n)

media = np.mean(x)
var_x = np.var(x, ddof=1)    # variância amostral (divide por n-1)
ep = np.sqrt(var_x) / np.sqrt(n)
variancia = var_x / n        # variância do estimador da média

print("E[X] estimada:", media, "(teórico 0.5)")
print("Var(X) estimada:", var_x, "(teórico 1/12 =", 1/12, ")")
print("Variância do estimador da média:", variancia)
print("IC 95% para E[X]:", media - 1.96*ep, media + 1.96*ep)
```
