**Passo 10 – Comparar estatísticas calculadas com valores teóricos**

```python
media_teorica_normal = distribuicoes['Normal']['params']['loc']
media_amostra_normal = distribuicoes['Normal']['estatisticas']['media']

media_teorica_exponencial = distribuicoes['Exponencial']['params']['scale']
media_amostra_exponencial = distribuicoes['Exponencial']['estatisticas']['media']

media_teorica_uniforme = (distribuicoes['Uniforme']['params']['low'] + distribuicoes['Uniforme']['params']['high']) / 2
media_amostra_uniforme = distribuicoes['Uniforme']['estatisticas']['media']

media_teorica_poisson = distribuicoes['Poisson']['params']['lam']
media_amostra_poisson = distribuicoes['Poisson']['estatisticas']['media']

media_teorica_binomial = distribuicoes['Binomial']['params']['n'] * distribuicoes['Binomial']['params']['p']
media_amostra_binomial = distribuicoes['Binomial']['estatisticas']['media']
```

```python
distribuicoes_nomes = ['Normal', 'Exponencial', 'Uniforme', 'Poisson', 'Binomial']
medias_teoricas = [media_teorica_normal, media_teorica_exponencial, media_teorica_uniforme, media_teorica_poisson, media_teorica_binomial]
medias_amostrais = [media_amostra_normal, media_amostra_exponencial, media_amostra_uniforme, media_amostra_poisson, media_amostra_binomial]

x = np.arange(len(distribuicoes_nomes))
largura = 0.35

plt.bar(x - largura/2, medias_teoricas, largura, label='Média teórica')
plt.bar(x + largura/2, medias_amostrais, largura, label='Média amostral')

plt.xticks(x, distribuicoes_nomes)
plt.title('Média teórica vs Média amostral')
plt.xlabel('Distribuição')
plt.ylabel('Média')
plt.legend()
plt.show()
```

Explicação: Por fim, calculei o valor teórico esperado da média de cada distribuição (usando as fórmulas de cada uma) e comparei com a média que realmente saiu da amostra gerada, usando um gráfico de barras. Os valores ficaram bem próximos, mostrando que a simulação fez sentido.
