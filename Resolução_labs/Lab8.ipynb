```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sqrt(2 / np.pi) * np.exp(-x**2 / 2)

def g(x):
    return np.exp(-x)

malha = np.linspace(0, 10, 100000)
razao = f(malha) / g(malha)

c = np.max(razao)
ponto_maximo = malha[np.argmax(razao)]

print("Ponto de máximo:", ponto_maximo)
print("c aproximado:", c)
print("c exato:", np.sqrt(2 * np.e / np.pi))
```

**1.2**

```python
n = 1000

amostra_abs = []
total_propostas = 0

for i in range(n):
    aceitou = False
    while aceitou == False:
        y = np.random.exponential(1)
        u = np.random.uniform(0, 1)
        total_propostas = total_propostas + 1
        if u <= np.exp(-(y - 1)**2 / 2):
            amostra_abs.append(y)
            aceitou = True
```

**1.3**

```python
amostra_normal = []

for x in amostra_abs:
    u = np.random.uniform(0, 1)
    if u < 0.5:
        amostra_normal.append(-x)
    else:
        amostra_normal.append(x)

amostra_numpy = np.random.normal(0, 1, n)

plt.hist(amostra_normal, bins=30, density=True, alpha=0.5, label="Aceitação-Rejeição")
plt.hist(amostra_numpy, bins=30, density=True, alpha=0.5, label="np.random.normal")
plt.xlabel("x")
plt.ylabel("Densidade")
plt.legend()
plt.show()
```

**1.4**

```python
mu = 2
sigma = 3

amostra_transformada = []
for z in amostra_normal:
    amostra_transformada.append(mu + sigma * z)
```

**1.5**

```python
amostra_numpy_mu = np.random.normal(mu, sigma, n)

plt.hist(amostra_transformada, bins=30, density=True, alpha=0.5, label="Transformação mu + sigma*Z")
plt.hist(amostra_numpy_mu, bins=30, density=True, alpha=0.5, label="np.random.normal")
plt.xlabel("x")
plt.ylabel("Densidade")
plt.legend()
plt.show()
```
