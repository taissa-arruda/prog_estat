**Exercício 1.1**

```python
import numpy as np
import matplotlib.pyplot as plt

D = 50

numero_bin = np.random.choice(2, size=D)
print(numero_bin)

resultado = 0

for i in range(0, D):
  resultado = resultado + numero_bin[i] * (1/2)**(1+i)

resultado
```

**Exercício 1.2**

```python
# numero_bin
potencias = np.arange(1, D+1)
print(resultado, np.sum((1/2)**potencias * numero_bin))
```

**Exercício 1.3**

```python
amostra_final = []

N = 10000
for i in range(N):

  numero_bin = np.random.choice(2, size=D)
  resultado = np.sum(((1/2)**potencias) * numero_bin)
  amostra_final.append(resultado)

amostra_uniforme = np.random.uniform(size=N)

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].hist(amostra_final, bins=30)
axs[0].set_title('Decomposição binária')

axs[1].hist(amostra_uniforme, bins=30)
axs[1].set_title('np.random.uniform')

plt.show()
```

**Exercício 1.4**

Comparando os dois histogramas, as amostras ficaram bem parecidas: nenhuma tem picos ou concentrações em alguma região específica do intervalo [0, 1], ambas ficaram espalhadas de forma equilibrada. Isso mostra que a decomposição binária produz números com comportamento compatível com uma distribuição uniforme, assim como o np.random.uniform. Faz sentido, já que cada dígito binário tem a mesma chance de ser 0 ou 1, então os números gerados a partir dessa sequência acabam se distribuindo de forma equilibrada por todo o intervalo, sem favorecer nenhuma região.

**Exercício 4.1**

```python
n = 10
p = 0.7
N = 1000

binomial_final = []
for i in range(N):
    u_bernoulli = np.random.uniform(size=n)
    bernoullis = (u_bernoulli < p).astype(int)
    binomial_final.append(np.sum(bernoullis))

binomial_final = np.array(binomial_final)
```

**Exercício 5.1**

```python
n = 20
probs = [0.10, 0.20, 0.30, 0.40]
cortes = np.cumsum(probs)  # [0.10, 0.30, 0.60, 1.00]

def f_multinomial(u):
    if u < cortes[0]:
        return 1
    elif u < cortes[1]:
        return 2
    elif u < cortes[2]:
        return 3
    else:
        return 4

u_lancamentos = np.random.uniform(size=n)
faces = np.array([f_multinomial(u) for u in u_lancamentos])

contagem = pd.Series(faces).value_counts().sort_index()
contagem
```

**Exercício 5.2**

```python
N = 1000
k = 4

amostra_multinomial = np.zeros((N, k))

for j in range(N):
    u_lancamentos = np.random.uniform(size=n)
    faces = np.array([f_multinomial(u) for u in u_lancamentos])
    contagem = pd.Series(faces).value_counts().reindex([1,2,3,4], fill_value=0)
    amostra_multinomial[j] = contagem.values

amostra_multinomial
```

```python
X1 = amostra_multinomial[:, 0]  # face 1
X2 = amostra_multinomial[:, 1]  # face 2

covariancia_amostral = np.mean(X1 * X2) - np.mean(X1) * np.mean(X2)

covariancia_teorica = -n * probs[0] * probs[1]

print(covariancia_amostral, covariancia_teorica)
```

**Interpretando o sinal negativo**

Isso faz sentido porque a soma de todas as faces é sempre fixa, igual a n (como confirmei no item 5). Então, se numa rodada a face 1 sair mais vezes do que o esperado, sobra menos espaço pras outras faces saírem — ou seja, quando X1 aumenta, X2 tende a diminuir, e vice-versa. Essa relação inversa entre as contagens é justamente o que a covariância negativa mostra: as faces "competem" pelo mesmo total fixo de n lançamentos.
